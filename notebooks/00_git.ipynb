```
OPERATOR   = "nikolai"                                  # <- your name
GIT_NAME   = "Nikolai"
GIT_EMAIL  = "you@example.com"                          # <- the email on your GitHub account
GITHUB     = "github.com/YOUR_USER/pei-landuse.git"    # <- private repo

from google.colab import drive, userdata
drive.mount('/content/drive')
import os, subprocess
ROOT = "/content/drive/MyDrive/pei_landuse"
REPO = f"{ROOT}/code/{OPERATOR}/pei-landuse"
TOKEN = userdata.get("GITHUB_TOKEN")
AUTH = f"https://{TOKEN}@{GITHUB}"

def git(*args, cwd=REPO):
    r = subprocess.run(["git", *args], cwd=cwd, capture_output=True, text=True)
    print((r.stdout + r.stderr).replace(TOKEN, "***"))
    return r.returncode == 0
```

```
if not os.path.exists(REPO):
    os.makedirs(os.path.dirname(REPO), exist_ok=True)
    git("clone", AUTH, REPO, cwd=os.path.dirname(REPO))
    git("remote", "set-url", "origin", f"https://{GITHUB}")   # don't keep the token in .git/config
git("config", "user.name", GIT_NAME); git("config", "user.email", GIT_EMAIL)
git("config", "core.fileMode", "false")                       # Drive messes with file permissions
subprocess.run("pip install -q nbstripout", shell=True)
subprocess.run("nbstripout --install", shell=True, cwd=REPO)   # strips notebook outputs on commit (map images!)
git("status", "--short", "--branch")
```

```
git("status", "--short", "--branch")
git("pull", "--ff-only", AUTH, subprocess.run(["git","branch","--show-current"],cwd=REPO,capture_output=True,text=True).stdout.strip())
```

```
BRANCH = "feat/my-change"
git("switch", "-c", BRANCH)
```

```
MESSAGE = "describe the change"
git("add", "-A")
git("status", "--short")
git("commit", "-m", MESSAGE)
git("push", AUTH, "HEAD")
```
